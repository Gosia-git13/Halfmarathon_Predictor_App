### Umieszczam dane w Digital Ocean Spaces

W pliku **.env** mam zdefiniowane 3 klucze
```
AWS_ACCESS_KEY_ID=...
AWS_SECRET_ACCESS_KEY=...
AWS_ENDPOINT_URL_S3=https://fra1.digitaloceanspaces.com
```

````python
# import bibliotek do obsługi kluczy:
from dotenv import load_dotenv
load_dotenv()
````

````python
# import potrzebnych bibliotek do komunikacji z Digital Ocean Spaces:
import boto3
import os

s3 = boto3.client(
    "s3"
)
````

````python
# definiuję zmienne:
BUCKET_NAME = "gotoitgosia13"
DATA_PREFIX = "halfmarathon/data/"
FILES = ["halfmarathon_wroclaw_2023__final.csv", "halfmarathon_wroclaw_2024__final.csv"]
````

````python
# wysyłam pliki do Digital Ocean Spaces:
for file in FILES:
    s3_path = f"{DATA_PREFIX}{file}"
    print(f"Wysyłam plik {file} do {s3_path}")
    s3.upload_file(
        Filename=file,
        Bucket=BUCKET_NAME,
        Key=s3_path,
    )
````

````python
# sprawdzam upload:
response = s3.list_objects_v2(Bucket=BUCKET_NAME, Prefix=DATA_PREFIX)

for obj in response["Contents"]:
    print(obj["Key"])
````
